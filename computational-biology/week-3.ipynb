```python
def validate_dna(dna_seq):
    seqm = dna_seq.upper()
    valid = seqm.count("A") + seqm.count("C") + seqm.count("G") + seqm.count("T")

    if valid == len(seqm):
        return True
    else:
        return False

validate_dna("atagagagatctcg")
validate_dna("ATAGAXTAGAT")
```

```python
def frequency(seq):
    dic = {}
    for s in seq.upper():
        if s in dic: 
            dic[s] += 1
        else:
            dic[s] = 1
    return dic

frequency("atagataactcgcatag")
```

```python
def reverse_complement(dna_seq):
    """ compute the reverse complement of the DNA sequence """
    assert validate_dna(dna_seq), "Invalid DNA sequence"
    comp = ""
    for c in dna_seq.upper():
        if c == "A":
            comp = "T" + comp
        elif c == "T":
            comp = "A" + comp
        elif c == "C":
            comp = "G" + comp
        elif c == "G":
            comp = "C" + comp
    return comp
```

```python
def translate_codon(cod):
    """Translates a codon into an aminoacid using an internal
    dictionary with the standard genetic code."""
    tc = {"GCT":"A", "GCC":"A", "GCA":"A", "GCG":"A",
          "TGT":"C", "TGC":"C",
          "GAT":"D", "GAC":"D",
          "GAA":"E", "GAG":"E",
          "TTT":"F", "TTC":"F",
          "GGT":"G", "GGC":"G", "GGA":"G", "GGG":"G",
          "CAT":"H", "CAC":"H",
          "ATA":"I", "ATT":"I", "ATC":"I",
          "AAA":"K", "AAG":"K",
          "TTA":"L", "TTG":"L", "CTT":"L", "CTC":"L", "CTA":"L", "CTG":"L",
          "ATG":"M", "AAT":"N", "AAC":"N",
          "CCT":"P", "CCC":"P", "CCA":"P", "CCG":"P",
          "CAA":"Q", "CAG":"Q",
          "CGT":"R", "CGC":"R", "CGA":"R", "CGG":"R", "AGA":"R", "AGG":"R",
          "TCT":"S", "TCC":"S", "TCA":"S", "TCG":"S", "AGT":"S", "AGC":"S",
          "ACT":"T", "ACC":"T", "ACA":"T", "ACG":"T",
          "GTT":"V", "GTC":"V", "GTA":"V", "GTG":"V",
          "TGG":"W",
          "TAT":"Y", "TAC":"Y",
          "TAA":"_", "TAG":"_", "TGA":"_"}
    if cod in tc: 
        return tc[cod]
    else: 
        return None
```

## Activity 1: Decode the DNA Sequence
### 1. Store The DNA Sequence

```python
# 3' - TAC CGA TTT ACC ACT - 5'
dna_template = "TACCGATTTACCACT"
```

### 2. Transcription: DNA template -> mRNA

```python
def transcription(dna):
    dna = dna.upper().replace(" ", "")

    if not validate_dna(dna):
        print("Invalid DNA sequence")

    mapping = {
        "A": "U", 
        "T": "A", 
        "C": "G", 
        "G": "C"
    }

    result = ""
    for base in dna:
        result += mapping[base]

    return result
```

### 3. Divide mRNA into codons

```python
# 3. Divide a sequence into codons 
def find_codons(sequence):
    codons = []

    for i in range(0, len(sequence), 3):
        codon = sequence[i : i + 3]

        if len(codon) == 3:
            codons.append(codon)

    return codons
```

### 4. Translate the codons

```python
def translate_sequence(codons):
    amino_acids = []

    for codon in codons:
        dna_codon = codon.replace("U", "T")
        amino_acid = translate_codon(dna_codon)

        if amino_acid == "_":
            break

        if amino_acid is None:
            raise ValueError(f"Unknown codon: {codon}")

        amino_acids.append(amino_acid)

    return amino_acids


mrna = transcription(dna_template)
codons = find_codons(mrna)
protein = translate_sequence(codons)

print("DNA:", dna_template)
print("mRNA:", mrna)
print("Codons:", " | ".join(codons))
print("Amino-acid sequence:", "-".join(protein))
```

## Final code challenge

### 1. Find an ORF

```python
def find_orf(coding_dna):
    start_position = coding_dna.find("ATG")

    if start_position == -1:
        return -1, None, None, "", []

    stop_codons = {"TAA", "TAG", "TGA"}
    stop_position = None
    stop_codon = None
    orf_codons = []

    # Read in groups of three, beginning at ATG.
    for position in range(start_position, len(coding_dna) - 2, 3):
        codon = coding_dna[position:position + 3]
        orf_codons.append(codon)

        if codon in stop_codons:
            stop_position = position
            stop_codon = codon
            break

    orf_sequence = "".join(orf_codons)

    return start_position, stop_position, stop_codon, orf_sequence, orf_codons
```

### 2. Build the analyzer

```python
def analyze_dna(dna, strand_type):
    dna = dna.upper().replace(" ", "")
    strand_type = strand_type.lower().strip()

    if not dna or not validate_dna(dna):
        print("Invalid DNA sequence. Use only A, T, C, and G.")
        return

    if strand_type == "template":
        # Input runs 3′ → 5′, as in Activity 1.
        mrna = transcription(dna)
    elif strand_type == "coding":
        # Input runs 5′ → 3′, as in Activity 2.
        mrna = dna.replace("T", "U")
    else:
        print("Strand type must be 'template' or 'coding'.")
        return

    codons = find_codons(mrna)

    # Convert mRNA to its matching coding DNA sequence for ORF analysis.
    coding_dna = mrna.replace("U", "T")
    start, stop, stop_codon, orf, orf_codons = find_orf(coding_dna)

    # Translate from ATG up to, but not including, the stop codon.
    protein_codons = []

    for codon in orf_codons:
        if codon in {"TAA", "TAG", "TGA"}:
            break

        protein_codons.append(codon.replace("T", "U"))

    protein = translate_sequence(protein_codons)
    valid_orf = start != -1 and stop is not None

    print("DNA sequence:", dna)
    print("mRNA sequence:", mrna)
    print("Codons:", " | ".join(codons))
    print("Protein sequence:", "-".join(protein) if protein else "None")
    print("Start codon position:", start if start != -1 else "Not found")
    print("Stop codon position:", stop if stop is not None else "Not found")
    print("Stop codon:", stop_codon if stop_codon else "Not found")
    print("ORF sequence:", orf if orf else "None")
    print("Valid ORF:", valid_orf)
```

### 3. Accept user input

```python
dna = input("Enter a DNA sequence: ")
strand_type = input("template or coding strand? ")

analyze_dna(dna, strand_type)
```
