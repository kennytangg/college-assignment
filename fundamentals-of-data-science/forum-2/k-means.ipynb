```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
```

### 1. Load datasets

```python
# Load data
digits = load_digits()
```

```python
X = digits.data          
y = digits.target        
```

### 2. Run K-Means

```python
kmeans = KMeans(n_clusters=10, init="k-means++", n_init=10, random_state=0, verbose=1)
kmeans.fit(X)
```

### 3. Visualize

```python
fig, axs = plt.subplots(2, 5, figsize=(8, 3))
for i, ax in enumerate(axs.flat):
    ax.imshow(kmeans.cluster_centers_[i].reshape(8, 8), cmap='gray')
    ax.set_title(f'Centroid {i}')
    ax.axis('off')
plt.suptitle('Cluster Centroids (K-Means)')
plt.tight_layout()
plt.show()
```

### 4. Movement and separation in 2D with PCA
This shows how data and centroids look in low dimensions

```python
X_pca = PCA(n_components=2).fit_transform(X)
kmeans_pca = KMeans(n_clusters=10, init="k-means++", n_init=10, random_state=0)
labels_pca = kmeans_pca.fit_predict(X_pca)
centroids_pca = kmeans_pca.cluster_centers_

plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels_pca, s=10, cmap='tab10')
plt.scatter(centroids_pca[:, 0], centroids_pca[:, 1], marker='x', s=200, color='black')
plt.title("K-Means Clusters and Centroids (PCA-reduced digits)")
plt.show()
```
