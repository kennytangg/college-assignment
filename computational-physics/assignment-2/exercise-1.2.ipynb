```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Create a grid of x and y values
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
X, Y = np.meshgrid(x, y)

# Calculate z values using f(x,y) = -x^2 + xy + y^2
Z = -X**2 + X*Y + Y**2

# Find min and max z values
z_min = np.min(Z)
z_max = np.max(Z)
print(f"Minimum z value: {z_min:.2f}")
print(f"Maximum z value: {z_max:.2f}")

# Create the 3D plot
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
surf = ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8, edgecolor='none')

# Set axis labels
ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('y', fontsize=12)
ax.set_zlabel('z = f(x,y)', fontsize=12)

# Set z-axis limits to include min and max values with some padding
z_range = z_max - z_min
ax.set_zlim(z_min - 0.1*z_range, z_max + 0.1*z_range)

# Set title
ax.set_title('Surface: f(x,y) = -x² + xy + y²', fontsize=14, pad=20)

# Add a colorbar
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=5)

plt.tight_layout()
plt.savefig('surface_plot.png', dpi=150, bbox_inches='tight')
plt.show()
```
